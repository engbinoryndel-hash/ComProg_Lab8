```python
class Book:
    def __init__(self, title, author, status="available"):
        self.title = title
        self.author = author
        self.status = status # [cite: 293]

class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title, author):
        new_book = Book(title, author)
        self.books.append(new_book)
        return f"Book '{title}' added successfully."

    def borrow_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if book.status == "available":
                    book.status = "borrowed" # [cite: 295, 296]
                    return f"You have successfully borrowed '{book.title}'."
                return f"Error: '{book.title}' is already borrowed." # [cite: 299]
        return "Error: Book not found." # [cite: 299]

    def return_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                book.status = "available" # [cite: 295]
                return f"You have successfully returned '{book.title}'."
        return "Error: Book not found."

    def display_books(self):
        print("\n--- Library Inventory ---")
        for book in self.books:
            print(f"Title: {book.title} | Author: {book.author} | Status: {book.status}") # [cite: 297]

# --- Example Run ---
my_library = Library()
num_books = int(input("Enter number of books to add: ")) # [cite: 292]

for _ in range(num_books):
    t = input("Enter title: ")
    a = input("Enter author: ")
    my_library.add_book(t, a)

my_library.display_books()
print(my_library.borrow_book(input("Enter title to borrow: ")))
my_library.display_books()
```

```python
class Employee:
    def __init__(self, name, hourly_rate, hours_worked):
        # Handle invalid inputs [cite: 305]
        if hourly_rate < 0 or hours_worked < 0:
            raise ValueError("Rate and hours cannot be negative.")
        
        self.name = name
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

    def calculate_salary(self):
        base_hours = min(self.hours_worked, 160)
        overtime_hours = max(0, self.hours_worked - 160)
        
        base_pay = base_hours * self.hourly_rate
        overtime_pay = overtime_hours * (self.hourly_rate * 1.5) # [cite: 303]
        total_salary = base_pay + overtime_pay
        
        return base_pay, overtime_pay, total_salary

# --- Example Run ---
employees = []
num_emp = int(input("Enter number of employees: ")) # [cite: 301]

for _ in range(num_emp):
    name = input("Name: ")
    rate = float(input("Hourly Rate: "))
    hours = float(input("Hours Worked: "))
    employees.append(Employee(name, rate, hours)) # [cite: 302]

print("\n--- Payroll Summary ---")
for emp in employees:
    base, oto, total = emp.calculate_salary()
    print(f"Name: {emp.name} | Hours: {emp.hours_worked} | Base: P{base:.2f} | Overtime: P{oto:.2f} | Total: P{total:.2f}") # [cite: 304]
```

```python
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock # [cite: 308]

class ShoppingCart:
    def __init__(self):
        self.items = {} # Stores product_name: quantity
        self.total_price = 0.0

    def add_to_cart(self, product, quantity):
        if quantity > product.stock: # [cite: 311, 313]
            return f"Error: Only {product.stock} items left in stock."
        
        if product.name in self.items:
            self.items[product.name] += quantity
        else:
            self.items[product.name] = quantity
            
        product.stock -= quantity # Update stock [cite: 311]
        self.total_price += product.price * quantity # [cite: 310]
        return f"Successfully added {quantity} {product.name}(s) to cart." # [cite: 312]

    def display_cart(self):
        print("\n--- Your Shopping Cart ---")
        for name, qty in self.items.items():
            print(f"Product: {name} | Quantity: {qty}")
        print(f"Total Price: P{self.total_price:.2f}") # [cite: 311]

# --- Example Run ---
# Predefined List [cite: 307]
catalog = {
    "1": Product("Laptop", 45000, 5),
    "2": Product("Mouse", 500, 20),
    "3": Product("Keyboard", 1200, 10)
}

my_cart = ShoppingCart()
print("1. Laptop (P45,000) | 2. Mouse (P500) | 3. Keyboard (P1,200)")
choice = input("Select product ID to add: ")
qty = int(input("Enter quantity: "))

if choice in catalog:
    print(my_cart.add_to_cart(catalog[choice], qty))
    my_cart.display_cart()
else:
    print("Invalid product selection.") # [cite: 313]
```
